```python
import random
import os
import itertools

testSetSize = 100
TestVectorSize = 178
file_path = "testPatterns"+str(testSetSize)+".txt"
TestSet = []
```

```python
import random

def generate_random_series(length, count):
    series_set = set()
    
    while len(series_set) < count:
        series = [random.randint(0, 1) for _ in range(length)]
        series_set.add(tuple(series))
    
    return [list(series) for series in series_set]

# Parameters
length = TestVectorSize  # Length of each series
count = testSetSize  # Number of unique series to generate

# Generate the series
random_series = generate_random_series(length, count)
random.shuffle(random_series)

# # Print the series
# for series in random_series:
#     print(series)
```
